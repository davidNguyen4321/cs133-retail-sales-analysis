```python
# Missing value count
for column in df.columns:
    nullCount = df[column].isna().sum()
    nullPercent = round(nullCount/len(df)*100, 3)
    print(f"{column}: {nullCount} missing ({nullPercent}%)")
```

```python
for column in df.columns:
    uniqueList = list(df[column].unique())
    uniqueLen = len(uniqueList)
    print(f"{column}: {uniqueLen} elements")
    print(uniqueList[:min(10, uniqueLen)]) #Prints the first ten values
```
